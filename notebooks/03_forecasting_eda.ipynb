# Forecasting EDA

This notebook explores demand patterns by date, store, category, and product so the forecasting pipeline is grounded in real seasonal and trend behavior.


```python
import sys
from pathlib import Path

project_root = Path.cwd().resolve()
for candidate in [project_root, *project_root.parents]:
    if (candidate / 'src').exists() and (candidate / 'scripts').exists():
        project_root = candidate
        break
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

import matplotlib
matplotlib.use('Agg')
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from scripts.generate_demo_data import ensure_demo_data
from src.forecasting.preprocessing import build_forecasting_base

ensure_demo_data()
base = build_forecasting_base()
print(base.head())
print('\nDataset summary:')
print(base[['sales', 'store_id', 'category', 'department']].describe())

by_store = base.groupby('store_id')['sales'].sum().sort_values(ascending=False)
print('\nSales by store:')
print(by_store)

by_category = base.groupby('category')['sales'].sum().sort_values(ascending=False)
print('\nSales by category:')
print(by_category)

monthly = base.assign(month_label=base['date'].dt.to_period('M').astype(str)).groupby('month_label')['sales'].sum()
plt.figure(figsize=(10, 4))
monthly.plot(kind='line')
plt.title('Monthly demand')
plt.tight_layout()
plt.show()

store_summary = base.groupby(['date', 'store_id'], as_index=False)['sales'].sum()
plt.figure(figsize=(10, 4))
sns.lineplot(data=store_summary, x='date', y='sales', hue='store_id', errorbar=None)
plt.title('Daily sales by store')
plt.tight_layout()
plt.show()
```
